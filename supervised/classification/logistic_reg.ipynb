## 1. The model

A linear model computes a score $z = \mathbf{w}\cdot\mathbf{x} + b$, which can be any real number. The **sigmoid** squeezes it into $(0, 1)$ so we can read it as a probability:

$$f_{\mathbf{w},b}(\mathbf{x}) = \sigma(z) = \frac{1}{1 + e^{-z}} = P(y=1 \mid \mathbf{x})$$

A score of 0 maps to probability 0.5. That is the decision boundary: predict class 1 when $z \ge 0$.
